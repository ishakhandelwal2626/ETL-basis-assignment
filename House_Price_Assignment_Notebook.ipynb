# House Price India Dataset Analysis
Assignment Submission Notebook

```

import pandas as pd
import numpy as np

df = pd.read_csv('House Price India.csv')
df.head()
```

```

# Dataset information
print(df.shape)
df.info()
df.describe()
```

## Q1 Average, Median and Standard Deviation of House Prices

```

avg_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()

print("Average:", avg_price)
print("Median:", median_price)
print("Standard Deviation:", std_price)
```

## Q2 Most Common Number of Bedrooms

```

df['number of bedrooms'].mode()
```

## Q3 Check Price Skewness

```

df['Price'].skew()
```

## Q4 Average Price for Each Number of Bedrooms

```

df.groupby('number of bedrooms')['Price'].mean().sort_index()
```

## Q5 Relationship Between Living Area and Price

```

df[['Area of the house(excluding basement)','Price']].corr()
```

## Q6 High Price House Anomalies

```

threshold = df['Price'].quantile(0.99)
df[df['Price'] > threshold].head()
```

## Q7 Average Price by Floors and Waterfront

```

df.groupby(['number of floors','waterfront present'])['Price'].mean()
```

## Q8 Minimum House Price by Condition

```

df.groupby('condition of the house')['Price'].min()
```

## Q9 Area/Zip Code with Highest Average Price

```

df.groupby('Postal Code')['Price'].mean().sort_values(ascending=False).head(10)
```
